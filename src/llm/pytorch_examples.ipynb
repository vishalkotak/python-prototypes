### 1D Convultion

```python
import torch

input_list = [1, 2, 3, 4, 5]
kernel_list = [1, 0, -1]

input_tensor = torch.tensor(input_list, dtype=torch.float32)
kernel_tensor = torch.tensor(kernel_list, dtype=torch.float32)
print(f"input tensor {input_tensor}")
print(f"kernel tensor {kernel_tensor}")

kernel_size = kernel_tensor.shape[0]
print(f"kernel size {kernel_size}")

stride = 1

# dimension = 0 Slide along the only dimension of the 1D tensor
# size = Each window should be of the size of the kernel
# step = How many elements to move for the next window
input_windows = input_tensor.unfold(0, kernel_size, stride)
print(f"input windows: {input_windows}")

# Perform element wise multiplication between windows and the kernel
# The kernel tensor [1, 0, -1] will be broadcast across the rows of 
# input_windows
multiplied = input_windows * kernel_tensor
print(f"multiplication between input window and kernel tensor {multiplied}")

output_tensor = torch.sum(multiplied, dim=0)
print(f"Output tensor: {output_tensor}")
```
