### Default values and function as first class objects

```python
def func(x, temp=[]):
    temp.append(x)
    return temp

print(func.__defaults__)
print(func.__name__)
print(func.__doc__)
print(id(func))
print(type(func))
```

### Class definition and instance attributes

```python
class Person:
    def __init__(self, name: str, age: float) -> None:
        self.name = name
        self.age = age

    def introduce(self) -> None:
        print(f"Hi, I'm {self.name} and I'm {self.age} years old")

person_1 = Person("Vishal", 32)
person_2 = Person("Akanksha", 33)
person_list = [person_1, person_2]
for person in person_list:
    person.introduce()
```

## Abstraction and Method overriding

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name: str) -> None:
        self.name = name

    @abstractmethod
    def sound(self) -> str:
        pass

class Dog(Animal):
    def __init__(self, name: str) -> None:
        super().__init__(name)

    def sound(self) -> str:
        return "Woof!"

class Cat(Animal):
    def __init__(self, name: str) -> None:
        super().__init__(name)

    def sound(self) -> str:
        return "Meow!"

animals = [Dog("Max"), Cat("Tom")]
for animal in animals:
    print(animal.sound())
```

### Encapsulation and Properties

```python
class BankAccount:
    def __init__(self) -> None:
        self.__balance = 0

    @property
    def balance(self) -> float:
        return self.__balance

    def deposit(self, amount) -> None:
        if amount <= 0:
            raise ValueError(f"Invalid amount {amount}")
        self.__balance += amount

    def withdraw(self, amount) -> None:
        if amount > self.__balance:
            raise ValueError(f"Amount {amount} more than the balance.")
        self.__balance -= amount

bank_account = BankAccount()
try:
    bank_account.deposit(0)
except Exception as e:
    print(f"{e}")
bank_account.deposit(100)
try:
    print(bank_account.__balance)
except Exception as e:
    print(f"{e}")
print(bank_account.balance)
try:
    bank_account.withdraw(110)
except Exception as e:
    print(f"{e}")
bank_account.withdraw(50)
```

### Polymorphism and Duck Typing

```python
class Rectangle:
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

class Circle:
    def __init__(self, radius: float):
        self.radius = radius

    def area(self) -> float:
        return 3.14 * (self.radius ** 2)

def calculate_total_area(shapes) -> float:
    total = 0
    for shape in shapes:
        total += shape.area()
    return total

shapes = [Rectangle(2, 3), Rectangle(3, 4), Circle(3)]
print(calculate_total_area(shapes))
```

### Special (Dunder) Methods & Operator Overloading

```python
class Vector:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    @property
    def x(self):
        return self.__x

    @property
    def y(self):
        return self.__y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        return self.x == other.x and self.y == other.y

vector1 = Vector(1, 1)
vector2 = Vector(2, 2)
print(vector1)
print(vector1 + vector2)
vector3 = Vector(1, 1)
print(vector1 == vector3)
```

### Abstract Classes and Interfaces

```python
from abc import ABC, abstractmethod

import math

class Shape(ABC):
    def __init__(self):
        pass

    @abstractmethod
    def get_area(self):
        pass

class Triangle(Shape):
    def __init__(self, base, height):
        self.__base = base
        self.__height = height

    @property
    def base(self):
        return self.__base

    @property
    def height(self):
        return self.__height

    @base.setter
    def base(self, new_base):
        self.__base = new_base

    @height.setter
    def height(self, new_height):
        self.__height = new_height
    
    def get_area(self):
        return 0.5 * self.base * self.height

class Ellipse(Shape):
    def __init__(self, radius):
        self.__radius = radius

    @property
    def radius(self):
        return self.__radius

    @radius.setter
    def radius(self, new_radius):
        self.__radius = new_radius

    def get_area(self):
        return math.pi * self.radius ** 2 

try:
    shape = Shape()
except Exception as e:
    print(f"{e}")
shapes = [Triangle(1, 1), Ellipse(1)]
triangle = Triangle(2, 3)
triangle.base = 1
for shape in shapes:
    print(shape.get_area())
```

### Composition and Delegation

```python
class Engine:
    
    def start(self):
        print(f"Engine started")

    def stop(self):
        print(f"Engine stopped")

class Wheel:
    pass

class Car:
    def __init__(self, num_wheels: int):
        # Car contains engine and wheel i.e. Composition
        self.__engine = Engine()
        self.__wheels = [Wheel() for _ in range(num_wheels)]

    @property
    def engine(self):
        return self.__engine

    @property
    def wheels(self):
        return self.__wheels

    def start(self):
        # Car offloads the start functionality to engine i.e. delegation
        self.engine.start()

    def stop(self):
        # Car offloads the stop functionality to engine i.e. delegation
        self.engine.stop()

    def describe(self):
        print(f"The car has {len(self.wheels)} wheels.")

car = Car(4)
car.start()
car.describe()
car.stop()
```

### Mini Library System

```python
from typing import List

class Item:
    def __init__(self, title: str, year: int) -> None:
        self.__title = title
        self.__year = year

    @property
    def title(self):
        return self.__title

    @property
    def year(self):
        return self.__year

    def describe(self) -> None:
        print(f"Title = {self.__title}, Year = {self.__year}.")

class Book(Item):
    def __init__(self, title: str, year: int, author: str) -> None:
        super().__init__(title, year)
        self.__author = author

    def __str__(self):
        return f"{self.__class__.__name__} - {self.title} - {self.year} - {self.__author}"

class Magazine(Item):
    def __init__(self, title: str, year: int, issue: int) -> None:
        super().__init__(title, year)
        self.__issue = issue

    def __str__(self):
        return f"{self.__class__.__name__} - {self.title} - {self.year} - {self.__issue}"

class DVD(Item):
    def __init__(self, title: str, year: int, duration: int) -> None:
        super().__init__(title, year)
        self.__duration = duration

    def __str__(self):
        return f"{self.__class__.__name__} - {self.title} - {self.year} - {self.__duration}"

class Member:
    def __init__(self, name: str, borrowed_items: List[Item]) -> None:
        self.__name = name
        self.__borrowed_items = borrowed_items

    def borrow(self, item: Item) -> None:
        if item:
            self.__borrowed_items.append(item)

    @property
    def borrowed_items(self):
        return self.__borrowed_items

    def return_item(self, item: Item) -> None:
        if len(self.__borrowed_items) == 0:
            raise ValueError(f"No items for removal")
        if item not in self.__borrowed_items:
            raise ValueError(f"Item {item} not present in borrowed items")
        self.__borrowed_items.remove(item)

items = [
         Book("Book", 2025, "Author"),
         Magazine("Magazine", 2025, "Issue"),
         DVD("DVD", 2025, 100)
        ]
member = Member("Member", items)
for item in member.borrowed_items:
    print(item)
add_item = Book("Book1", 2025, "Author")
print(f"----------- Adding Item --------------")
member.borrow(add_item)
for item in member.borrowed_items:
    print(item)
print(f"----------- Removing Item Failed --------------")
try:
    remove_item = Book("Book2", 2025, "Author")
    member.return_item(remove_item)
except ValueError as e:
    print(f"{e}")
print(f"----------- Removing Item Success --------------")
member.return_item(add_item)
for item in member.borrowed_items:
    print(item)
        
```
