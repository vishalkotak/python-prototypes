```python
word_vectors = [[1, 1, 1], [2, 2, 2], [4, 4, 4], [5, 5, 5]]
given_vector = [0, 1, 2]

import heapq
min_heap = []

for word_vector in word_vectors:
    distance = 0
    for i in range(len(given_vector)):
        distance += (abs(word_vector[i] - given_vector[i]) ** 2)
    heapq.heappush(min_heap, (distance, word_vector))

no_of_closest_vector_required = 2
result = []
for i in range(no_of_closest_vector_required):
    _, min_distance_vector = heapq.heappop(min_heap)
    result.append(min_distance_vector)

print(result)
```
