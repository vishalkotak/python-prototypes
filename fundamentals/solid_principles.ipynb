# SOLID Principles

### Single Responsibility Principle

```python
# Multiple responsibilities in the same class
class Report:
    def generate(self):
        print(f"Generated report")

    def format(self):
        print(f"Formatting report")

    def save(self):
        print(f"Saving report")

# Every class has its own responsibility
class Report:
    def generate(self):
        print(f"Generated report")

class Formatter:
    def format(self):
        print(f"Formatting report")

class Writer:
    def save(self):
        print(f"Saving report")
```

### Open/Closed Principle

```python
class PaymentProcessor:
    def process(self, payee_details):
        print(f"Process card")

from abc import ABC, abstractmethod
class PaymentProcesor(ABC):
    @abstractmethod
    def process(self, payee_details):
        pass

class CreditCardProcessor(PaymentProcesor):
    def process(self, payee_details):
        pass

class PayPalProcessor(PaymentProcessor):
    def process(self, payee_details):
        pass

class BitcoinProcessor(PaymentProcesor):
    def process(self, payee_details):
        pass
```

### Liskov Substitution Principle

```python
class Bird:
    def fly(self):
        print(f"Bird is flying.")

class Penguin(Bird):
    pass

class Sparrow(Bird):
    pass

bird = Bird()
bird.fly()

# Penguin should not be ably to fly
# Can't replace bird with penguin
penguin = Penguin()
penguin.fly()

# Sparrow should be able to fly
sparrow = Sparrow()
sparrow.fly()

print("----- After Laskov Principle ------")

class Bird:
    def sound(self):
        print(f"Sound")

class Flyable:
    def fly(self):
        print(f"Fly")

class NonFlyable:
    def walk(self):
        print(f"Walk")

class Sparrow(Bird, Flyable):
    pass

class Penguin(Bird, NonFlyable):
    pass

penguin = Penguin()
penguin.walk()
try:
    penguin.fly()
except AttributeError as e:
    print(f"{e}")
```
